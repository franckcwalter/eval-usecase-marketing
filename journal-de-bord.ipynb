### 📝 Observations / Difficultés
- les critères de succès n'ont pas été exprimés par le sujet, pas de mention d'objectifs chiffrés attendus par le client, donc il n'a pas été aisé d'imaginer un critère chiffré pour ce client
- de même, sans avoir encore entraîné de modèle, comment imaginer de que qu'on peut attendre d'un modèle ? comment fixer un critère raisonnable et atteignable, et non simplement un critère idéal
- le succès in fine de la campagne dépend du prix d'un appel et de ce que rapporte un client qui souscrit effectivement au contrat, des informations qui ne sont pas disponilbes ; en effet, en fonction du coût des appels et de combien rapporte un contrat, et en fonction des ressources matérielles effectives de la banque, le problème peut-être posé de différentes manières : est-ce la banque ne peut ou ne peut appelquer que 50% des clients, et donc on cherche à cibler les 50% les plus susceptibles de sourscire ? ou est-ce qu'on cherche à optimiser le retour sur investissment avec des ressources illimitées et dans ce cas on peut appeler jusqu'à 80% des clients pour être sûr de rater le moins possible de clients potentiels ?
- des recherches documentaires m'ont permis d'estimer ces informations pour m'aider à estimer un critère de succès
