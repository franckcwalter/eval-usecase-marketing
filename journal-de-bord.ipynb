### 📝 Observations / Difficultés
- les critères de succès n'ont pas été exprimés par le sujet, pas de mention d'objectifs chiffrés attendus par le client, donc il n'a pas été aisé d'imaginer un critère chiffré pour ce client

## 📅 Journal de bord – Jour 6 (30/09/2026)

### 🔧 Étapes / Actions réalisées
- choix des familles de modèles et des modèles candidats
- entraînement de la régression logistique, de l'arbre de décision et de la forêt aléatoire sur les huit scénarios
- validation croisée et comparaison du rappel à 50% des clients sélectionnés et du ROC-AUC
- comparaison des résultats à 25% et 75% des clients sélectionnés
- tests et optimisation des hyperparamètres de l'arbre de décision S5
- comparaison de la baseline et du réglage retenu sur plusieurs découpages de validation croisée
- conclusion commune et choix de la régression logistique S5
- évaluation finale sur le jeu de test et synthèse de la modélisation

### 📝 Observations / Difficultés
- choisir les métriques en fonction de l'objectif métier : retrouver au moins 80% des souscripteurs en appelant 50% des clients
- les modèles les mieux classés changent selon le nombre de clients que la banque peut appeler
- la régression logistique arrive en tête à 50% d'appels, les arbres arrivent en tête à 25% et 75%, avec des résultats proches
- difficulté : décider si les petits gains de l'arbre justifient de changer de modèle
- le réglage des hyperparamètres de l'arbre n'apporte pas de gain stable face à la baseline
- choix de la régression logistique S5 pour le compromis entre performance, simplicité et explicabilité
- l'objectif est atteint sur le test : 82,33% des souscripteurs retrouvés en appelant 50% des clients
- les résultats sur le test restent proches de la validation croisée

### ✅ Prochaines étapes

### 📝 Observations / Difficultés
- les scénarios S5 et S8 ont des rappels proches à 50% d’appels ; S5 conserve de meilleurs résultats quand le volume d’appels varie

- le client le mieux classé a une probabilité prédite de 97,8% mais n’a pas souscrit
- le retrait des variables sensibles n’empêche pas les écarts de rappel entre profils : 68,0% chez les ouvriers et 70,5% chez les 45–54 ans, contre 82,3% sur l’ensemble
- les 30% de clients les mieux classés regroupent déjà 72,1% des souscripteurs ; les tranches suivantes ont des taux de souscription faibles et proches; ça pousse à proposer de se contenter d'appeler 30% des clients pour optimiser
